## PASO 0: Entendiendo las Vías de Acceso a los Datos S3 (Simple Storage Service)

Antes de realizar cualquier consulta, es fundamental comprender cómo están organizados los datos en la infraestructura del IDEAM y por qué la elección del formato transformará su experiencia de usuario.

El Instituto ha dispuesto la información hidrológica en un sistema de almacenamiento de objetos de alta velocidad llamado **S3 (Simple Storage Service)**. Como usuario, puede interactuar con este almacenamiento de dos formas distintas según su necesidad técnica y el volumen de datos que vaya a procesar.

A continuación, se explican las dos opciones disponibles para que elija la que mejor se adapte a su flujo de trabajo:

### 📑 Opción 1: Descarga de Archivos Temporales en Formato CSV
* **¿Cómo funciona?** El script se conecta al almacenamiento S3 (Simple Storage Service), busca el archivo de texto plano (`.csv`) correspondiente a su consulta, lo descarga completo a una ruta local o temporal de su máquina/servidor, y luego lo lee en memoria.
* **Ventajas:** Obtienes una copia física del archivo original estructurado por estaciones o variables, el cual puede guardar o abrir después en herramientas tradicionales si así lo requiere.
* **Desventajas:** Es la opción más lenta y pesada. Los archivos CSV no están indexados ni comprimidos eficientemente; descargar años de historia hidrológica de múltiples estaciones implicará transferir muchos Megabytes por la red, tardando más tiempo en procesar.
* **¿Cuándo usarla?:** Cuando necesite estrictamente el archivo plano original para auditoría o para guardarlo localmente en una ruta específica de su infraestructura.

### ⚡ Opción 2: Acceso Directo y Lectura Columnar en Formato Parquet
* **¿Cómo funciona?** El script no descarga archivos completos a su disco. Se conecta directamente a un set de datos optimizado en formato **Parquet** almacenado en el S3 (Simple Storage Service). Usando librerías especializadas en Python, lee "en la nube" únicamente las columnas y filas que coinciden con sus filtros (por ejemplo, solo la variable *Nivel* para un año específico).
* **Ventajas:** **Velocidad ultra rápida y máxima eficiencia.** Parquet es un formato de almacenamiento columnar y comprimido. Si un archivo contiene 20 variables pero solo pide *Nivel*, el sistema solo lee el fragmento del archivo que contiene los niveles. No hay descargas innecesarias de archivos gigantes, ahorrando red, memoria y tiempo de espera de minutos a solo fracciones de segundo.
* **Desventajas:** No genera un archivo de texto plano "legible a simple vista" en su disco local de manera inmediata; los datos van directo a la memoria del Notebook para su análisis.
* **¿Cuándo usarla?:** Siempre que sea posible. Es la opción recomendada para analítica de datos, creación de gráficos, modelos predictivos y consultas históricas masivas.

---

### 📊 Tabla Comparativa de Decisión

| Característica | Opción 1: CSV (.csv) | Opción 2: Parquet (.parquet) |
| :--- | :--- | :--- |
| **Ubicación del procesamiento** | Descarga local y luego procesa | Lectura directa en el almacenamiento S3 |
| **Velocidad de consulta** | 🐌 Más lenta (Depende del tamaño del archivo completo) | ⚡ Muy rápida (Solo lee lo que necesitas) |
| **Consumo de red / Ancho de banda** | Alto (Descarga todo el archivo) | Mínimo (Transfiere solo los datos filtrados) |
| **Uso ideal** | Respaldos físicos, uso en herramientas externas | Analítica en Python, reportes rápidos, series largas |

## PASO 6: Refinamiento por Características Técnicas de la Estación

Una vez delimitada la zona geográfica de interés, es fundamental filtrar las estaciones según sus especificaciones técnicas de infraestructura. Por ejemplo, si está analizando caudales o niveles, le interesará concentrarse únicamente en estaciones de categoría *Limnigráfica* o *Limnimétrica* que se encuentren operativas (*Activas*).

En esta celda, usted cruzará tres dimensiones técnicas fundamentales empleando las opciones oficiales del repositorio del IDEAM:

1. **`Clase`:** Clase de estación (Ej: `"Hidrológica"`, `"Meteorológica"`)
2. **`Categoria`:** Categoría de estación (Ej: `"Limnigráfica"`, `"Limnimétrica"`, `"Pluviométrica"`, `"Climatológica Principal"`).
3. **`Estado`:** Condición operativa actual (`"Activa"`, `"En Mantenimiento"`, `"Suspendida"`).
4. **`Tecnologia`:** Tipo de tecnología (`"Convencional"`, `"Automática con Telemetría"`, `"Automática sin Telemetría"`).

---
**Instrucciones:** Escriba textualmente las variables de configuración con los atributos técnicos que requiera su consulta. Si no desea restringir alguna de las dimensiones, escriba simplemente `"TODOS"` en mayúsculas sostenidas.

```python
# ==============================================================================
# PASO 6: REFINAMIENTO TÉCNICO DE INFRAESTRUCTURA (FILTRO DE CONTROL DE TOKENS)
# ==============================================================================

# ==============================================================================
# CONFIGURACIÓN DE ATRIBUTOS TÉCNICOS POR EL USUARIO
# ==============================================================================

# Opciones: "Limnigráfica", "Limnimétrica", "Pluviométrica", "Climatológica Principal", "TODOS"
filtro_categoria  = "Limnimétrica"

# Opciones: "Activa", "En Mantenimiento", "Suspendida", "TODOS"
filtro_estado     = "Activa"

# Opciones sugeridas: "Automática", "Convencional", "TODOS"
# (Si elige 'Automática', incluirá también las estaciones híbridas 'Automática + Convencional')
filtro_tecnologia = "Convencional"

# ==============================================================================

# --- Función auxiliar para remover tildes y caracteres especiales (No modificar) ---
def remover_tildes(texto):
    if not isinstance(texto, str):
        return str(texto)
    remplazos = (("Á", "A"), ("É", "E"), ("Í", "I"), ("Ó", "O"), ("Ú", "U"))
    t = texto.upper().strip()
    for original, reemplazo in remplazos:
        t = t.replace(original, reemplazo)
    return t

# --- Lógica de Filtrado Secuencial Cruzado (No modificar) ---

if 'df_estaciones_geo' in locals():
    # Partimos del subconjunto geográfico consolidado en el Paso 5C
    df_refinado = df_estaciones_geo.copy()

    # ---- 1. Filtrado por Categoría ----
    cat_usuario = remover_tildes(filtro_categoria)
    if cat_usuario != "TODOS":
        df_refinado['tmp_cat'] = df_refinado['Categoria'].apply(remover_tildes)
        df_refinado = df_refinado[df_refinado['tmp_cat'] == cat_usuario]
        df_refinado = df_refinado.drop(columns=['tmp_cat'])

    # ---- 2. Filtrado por Estado Operativo ----
    est_usuario = remover_tildes(filtro_estado)
    if est_usuario != "TODOS":
        df_refinado['tmp_est'] = df_refinado['Estado'].apply(remover_tildes)
        df_refinado = df_refinado[df_refinado['tmp_est'] == est_usuario]
        df_refinado = df_refinado.drop(columns=['tmp_est'])

    # ---- 3. Filtrado Inteligente por Tecnología (Resolución de Híbridos) ----
    tec_usuario = remover_tildes(filtro_tecnologia)
    if tec_usuario != "TODOS":
        df_refinado['tmp_tec'] = df_refinado['Tecnologia'].apply(remover_tildes)

        # Estrategia de Tokenización: Extraemos la raíz de la palabra seleccionada por el usuario
        # Si el usuario digita "Automática con Telemetría" o simplemente "Automática", el token principal será "AUTOMATICA"
        if "AUTOMATICA" in tec_usuario:
            token_busqueda = "AUTOMATICA"
        elif "CONVENCIONAL" in tec_usuario:
            token_busqueda = "CONVENCIONAL"
        else:
            token_busqueda = tec_usuario # Resguardo por si digita otro texto

        # El filtro busca el token dentro de la celda.
        # Así, un registro como "AUTOMÁTICA CON TELEMETRÍA, CONVENCIONAL" pasa la prueba si se busca "Automática" O "Convencional"
        df_refinado = df_refinado[df_refinado['tmp_tec'].str.contains(token_busqueda, na=False)]
        df_refinado = df_refinado.drop(columns=['tmp_tec'])

    # ---- 4. Despliegue de Resultados Definitivos ----
    if not df_refinado.empty:
        # Extraemos y blindamos la lista de códigos de estaciones final para el cruce de observaciones
        LISTA_ESTACIONES_TARGET = df_refinado['Codigo'].unique().tolist()

        print(f"✅ REFINAMIENTO TÉCNICO COMPLETADO CON ÉXITO (MÉTODO TOKENIZADO)")
        print("-" * 115)
        print(f"📊 Criterios Activos: Categoría = '{filtro_categoria.upper()}' | Estado = '{filtro_estado.upper()}' | Criterio Tecnología = '{filtro_tecnologia.upper()}'")
        print(f"🎯 Grupo de Control: Se han consolidado {len(LISTA_ESTACIONES_TARGET)} estaciones objetivo para la consulta histórica.")
        print("-" * 115)

        # Renderizado de la tabla final resumen en mayúsculas sostenidas para presentación
        columnas_resumen = ['Codigo', 'Nombre', 'Categoria', 'Tecnologia', 'Estado', 'Municipio', 'Corriente']
        df_vista_final = df_refinado[columnas_resumen].copy()
        for col in df_vista_final.columns:
            df_vista_final[col] = df_vista_final[col].astype(str).str.upper()

        print("\n📋 INVENTARIO DEFINITIVO DE ESTACIONES SELECCIONADAS:")
        print(df_vista_final.to_string(index=False))
    else:
        print("❌ No se encontraron estaciones en esta zona que cumplan simultáneamente con las características técnicas seleccionadas.")
        print("💡 Sugerencia: Flexibilice sus criterios (por ejemplo, cambie el filtro de Tecnología o Estado a 'TODOS') y vuelva a ejecutar.")
else:
    print("❌ ERROR DE SECUENCIA: No se detecta el subconjunto geográfico preliminar en memoria.")
    print("Por favor, ejecute primero la celda del Paso 5C.")
```

## PASO 7: Definición del Rango Temporal de Consulta (Filtro Desde / Hasta)

Las observaciones históricas de las variables hidrológicas representan el volumen de datos más masivo dentro del repositorio **S3 (Simple Storage Service)**. Para optimizar los tiempos de descarga y el uso de memoria, es una buena práctica delimitar la ventana de tiempo que desea analizar.

Usted puede configurar este filtro de dos maneras:
1. **Rango Acotado (Recomendado para actualizaciones):** Defina fechas específicas de inicio y fin (Formato: `AAAA-MM-DD`) para traer únicamente los registros de un período particular.
2. **Histórico Total (Primera ejecución):** Puede dejar las fechas abiertas (o un rango muy amplio como `1900-01-01` hasta la fecha actual) para extraer toda la serie temporal disponible para el grupo de estaciones seleccionado.

---
**Instrucciones:** 1. Modifique la variable `fecha_desde` con la fecha inicial de la consulta.
2. Modifique la variable `fecha_hasta` con la fecha final de la consulta.
3. El sistema validará automáticamente que el formato sea correcto (`AAAA-MM-DD`) y que la fecha inicial sea menor o igual a la final antes de dar paso a la extracción física de las observaciones.

```python
# ==============================================================================
# PASO 7: CONFIGURACIÓN Y VALIDACIÓN DEL RANGO DE FECHAS DE ESTUDIO
# ==============================================================================
import datetime

# ==============================================================================
# MODIFIQUE AQUÍ: Defina la ventana de tiempo para la extracción de observaciones
# Formato requerido: "AAAA-MM-DD" (Año-Mes-Día)
# ==============================================================================
fecha_desde = "2010-01-01"  # Modifique por la fecha inicial de su análisis
fecha_hasta = "2025-01-01"  # Modifique por la fecha final (o deje la fecha actual)
# ==============================================================================

# --- Lógica Institucional de Validación de Tiempos (No modificar) ---

def validar_formato_fecha(cadena_fecha):
    """Valida y convierte una cadena de texto a un objeto datetime.date"""
    try:
        return datetime.datetime.strptime(cadena_fecha.strip(), "%Y-%m-%d").date()
    except ValueError:
        return None

# Validamos que existan estaciones seleccionadas en el paso anterior
if 'LISTA_ESTACIONES_TARGET' in locals() and len(LISTA_ESTACIONES_TARGET) > 0:

    # Parseo y validación de las estructuras de fecha ingresadas por el usuario
    f_inicio = validar_formato_fecha(fecha_desde)
    f_final  = validar_formato_fecha(fecha_hasta)

    if f_inicio and f_final:
        # Validación de consistencia cronológica
        if f_inicio <= f_final:

            # Fijación de variables globales de control temporal en memoria del entorno
            FECHA_FILTRO_INICIO = f_inicio.strftime("%Y-%m-%d")
            FECHA_FILTRO_FINAL  = f_final.strftime("%Y-%m-%d")

            print(f"✅ RANGO TEMPORAL CONFIGURADO Y VALIDADO CON ÉXITO")
            print("-" * 80)
            print(f"📅 Fecha Inicial Activa: {FECHA_FILTRO_INICIO}")
            print(f"📅 Fecha Final Activa:   {FECHA_FILTRO_FINAL}")
            print(f"📡 Estaciones Objetivo en Memoria: {len(LISTA_ESTACIONES_TARGET)} estaciones.")
            if 'ETIQUETA_ESTUDIO' in locals():
                print(f"🏷️ Variable / Etiqueta Técnica Activa: '{ETIQUETA_ESTUDIO}'")
            print("-" * 80)
            print("\n🚀 El entorno de control está completamente blindado y listo.")
            print("Proceda a la siguiente celda para ejecutar la consulta masiva y la extracción física de datos.")

        else:
            print("❌ ERROR CRONOLÓGICO: La 'fecha_desde' no puede ser posterior a la 'fecha_hasta'.")
            print(f"Revisar: {fecha_desde} > {fecha_hasta}. Corrija los valores y vuelva a ejecutar.")
    else:
        print("❌ ERROR DE FORMATO: Una o ambas fechas no cumplen con la estructura estándar requerida.")
        print(f"Asegúrese de emplear estrictamente el formato 'AAAA-MM-DD' con guiones.")
        print(f"Valores ingresados actualmente ➡️ Desde: '{fecha_desde}' | Hasta: '{fecha_hasta}'")
else:
    print("❌ ERROR DE SECUENCIA: No se detecta la lista de estaciones objetivo en memoria.")
    print("Por favor, asegúrese de haber completado y ejecutado con éxito el Paso 6 antes de fijar las fechas.")
```

## PASO 8A: Extracción y Consolidación Local (Entorno Jupyter)

A diferencia de los pasos anteriores donde hemos ejecutado procesos en la nube, este bloque está diseñado para integrarse directamente con su entorno de desarrollo local.

---
### ⚠️ Requisito Técnico para la Ejecución Local
Para que el script pueda gestionar la descarga y escritura de sus archivos en su disco duro de manera autónoma, es indispensable que cuente con **Jupyter Notebook** (o JupyterLab) instalado en su computadora junto con una distribución de **Python**.

**¿Por qué este paso es multiplataforma?**
Hemos implementado una arquitectura basada en el módulo `pathlib`, lo que significa que el código es **100% independiente del Sistema Operativo**:
* **Si usa Windows:** El script identificará automáticamente sus rutas (ej: `C:\Usuarios\...`).
* **Si usa macOS o Linux:** El script gestionará sus directorios nativos sin errores.

**¿Qué realiza este paso?**
1. **Conexión Directa:** Establece un puente entre su máquina local y el S3 del IDEAM.
2. **Escritura Nativa:** Crea, ordena y guarda cada archivo `.csv` en la carpeta que usted haya definido en su equipo.
3. **Autonomía:** Al correr este proceso localmente, usted evita tiempos de latencia de red en la nube y mantiene un respaldo físico inmediato de sus series de tiempo.

---
**Instrucciones:** Asegúrese de haber instalado las librerías necesarias (`pip install pandas pyarrow s3fs`). Una vez verificado su entorno local, ejecute la celda a continuación. El sistema le notificará el progreso de la escritura en tiempo real.

```python
# ==============================================================================
# PASO 8A: EXTRACTOR USANDO JUPYTER LOCAL PARA MULTIPLES SISTEMAS OPERATIVOS
# ==============================================================================
import os
from pathlib import Path
import pandas as pd

var_etiqueta   = globals().get('ETIQUETA_ESTUDIO', locals().get('ETIQUETA_ESTUDIO'))
var_estaciones = globals().get('LISTA_ESTACIONES_TARGET', locals().get('LISTA_ESTACIONES_TARGET'))
var_f_inicio   = globals().get('FECHA_FILTRO_INICIO', locals().get('FECHA_FILTRO_INICIO'))
var_f_final    = globals().get('FECHA_FILTRO_FINAL', locals().get('FECHA_FILTRO_FINAL'))
var_ruta_pc    = globals().get('RUTA_DESTINO_LOCAL', locals().get('RUTA_DESTINO_LOCAL'))
opciones_env   = globals().get('opciones_almacenamiento')

ts_inicio = pd.to_datetime(var_f_inicio)
ts_final  = pd.to_datetime(var_f_final)

# --- RESOLUCIÓN ABSOLUTA DE RUTA EN MAC ---
# pathlib.Path resuelve el directorio real expandiendo cualquier restricción del sistema
directorio_destino = Path(var_ruta_pc.strip()).resolve()
directorio_destino.mkdir(parents=True, exist_ok=True)

print("🚀 INICIANDO EXTRACCIÓN CON ESCRITURA FÍSICA DIRECTA EN DISCO...")
print("-" * 95)
print(f"📁 Carpeta Real Verificada: {directorio_destino}")
print("-" * 95)

estaciones_exportadas = 0
total_filas_guardadas = 0

for cod_estacion in var_estaciones:
    id_estacion = str(cod_estacion).strip().zfill(10)
    ruta_particion = f"s3://s3-estacionesideam/observaciones/historicos/parquet/label={var_etiqueta}/id_estacion={id_estacion}"

    try:
        df_particion = pd.read_parquet(ruta_particion, storage_options=opciones_env)

        if df_particion.empty:
            continue

        df_particion['label'] = var_etiqueta
        df_particion['id_estacion'] = id_estacion
        df_particion['fecha'] = pd.to_datetime(df_particion['fecha'])

        mascara_fechas = (df_particion['fecha'] >= ts_inicio) & (df_particion['fecha'] <= ts_final)
        df_estacion_filtrado = df_particion[mascara_fechas].copy()

        if not df_estacion_filtrado.empty:
            columnas_ordenadas = ['label', 'id_estacion', 'fecha', 'valor', 'nivel_aprobacion']
            df_estacion_filtrado = df_estacion_filtrado[columnas_ordenadas]
            df_estacion_filtrado = df_estacion_filtrado.sort_values(by='fecha').reset_index(drop=True)

            # Formateo visual strings
            df_estacion_filtrado['label'] = df_estacion_filtrado['label'].astype(str).str.upper()
            df_estacion_filtrado['nivel_aprobacion'] = df_estacion_filtrado['nivel_aprobacion'].astype(str).str.upper()

            # --- CONSTRUCCIÓN DE RUTA NATIVA MAC ---
            nombre_archivo = f"{var_etiqueta.upper()}-{id_estacion}.csv"
            ruta_archivo_fisico = directorio_destino / nombre_archivo

            # Forzamos la escritura con flush de sistema
            df_estacion_filtrado.to_csv(ruta_archivo_fisico, index=False, encoding='utf-8')

            estaciones_exportadas += 1
            total_filas_guardadas += len(df_estacion_filtrado)
            print(f"✅ ARCHIVO ESCRITO EN DISCO ➡️ {nombre_archivo} ({len(df_estacion_filtrado)} filas)")

    except Exception:
        continue

print("-" * 95)
print(f"🏁 PROCESO TERMINADO. Archivos físicos reales creados: {estaciones_exportadas}")
print(f"📊 Total registros transferidos: {total_filas_guardadas}")
```

```python
"""
# ==============================================================================
# PASO 8B: EXTRACTOR MASIVO OPTIMIZADO PARA GOOGLE COLAB CON DESCARGA A PC
# ==============================================================================
import os
from pathlib import Path
import pandas as pd
from google.colab import files  # Librería nativa para interactuar con tu PC real

var_etiqueta   = globals().get('ETIQUETA_ESTUDIO', locals().get('ETIQUETA_ESTUDIO'))
var_estaciones = globals().get('LISTA_ESTACIONES_TARGET', locals().get('LISTA_ESTACIONES_TARGET'))
var_f_inicio   = globals().get('FECHA_FILTRO_INICIO', locals().get('FECHA_FILTRO_INICIO'))
var_f_final    = globals().get('FECHA_FILTRO_FINAL', locals().get('FECHA_FILTRO_FINAL'))
opciones_env   = globals().get('opciones_almacenamiento')

ts_inicio = pd.to_datetime(var_f_inicio)
ts_final  = pd.to_datetime(var_f_final)

# En Colab, creamos una carpeta temporal dentro de la máquina virtual de Google
directorio_temporal_colab = Path("./datos_ideam_temporales")
directorio_temporal_colab.mkdir(parents=True, exist_ok=True)

print("🚀 INICIANDO EXTRACCIÓN EN LA NUBE DE COLAB...")
print("-" * 95)

estaciones_exportadas = 0

for cod_estacion in var_estaciones:
    id_estacion = str(cod_estacion).strip().zfill(10)
    ruta_particion = f"s3://s3-estacionesideam/observaciones/historicos/parquet/label={var_etiqueta}/id_estacion={id_estacion}"

    try:
        df_particion = pd.read_parquet(ruta_particion, storage_options=opciones_env)
        if df_particion.empty:
            continue

        df_particion['label'] = var_etiqueta
        df_particion['id_estacion'] = id_estacion
        df_particion['fecha'] = pd.to_datetime(df_particion['fecha'])

        mascara_fechas = (df_particion['fecha'] >= ts_inicio) & (df_particion['fecha'] <= ts_final)
        df_estacion_filtrado = df_particion[mascara_fechas].copy()

        if not df_estacion_filtrado.empty:
            columnas_ordenadas = ['label', 'id_estacion', 'fecha', 'valor', 'nivel_aprobacion']
            df_estacion_filtrado = df_estacion_filtrado[columnas_ordenadas]
            df_estacion_filtrado = df_estacion_filtrado.sort_values(by='fecha').reset_index(drop=True)

            # Formateo de strings
            df_estacion_filtrado['label'] = df_estacion_filtrado['label'].astype(str).str.upper()
            df_estacion_filtrado['nivel_aprobacion'] = df_estacion_filtrado['nivel_aprobacion'].astype(str).str.upper()

            # Guardamos el archivo temporalmente en el servidor de Google
            nombre_archivo = f"{var_etiqueta.upper()}-{id_estacion}.csv"
            ruta_temporal_archivo = directorio_temporal_colab / nombre_archivo
            df_estacion_filtrado.to_csv(ruta_temporal_archivo, index=False, encoding='utf-8')

            print(f"📦 Procesado en la nube ➡️ {nombre_archivo}")

            # --- EL TRUCO PARA TU MAC ---
            # Le ordenamos al navegador que descargue el archivo guardado en Google a tu PC real
            print(f"⬇️ Transfiriendo '{nombre_archivo}' a su computador local...")
            files.download(str(ruta_temporal_archivo))

            estaciones_exportadas += 1

    except Exception:
        continue

print("-" * 95)
print(f"🏁 PROCESO TERMINADO. Archivos enviados a tu carpeta de Descargas local: {estaciones_exportadas}")
"""
```
